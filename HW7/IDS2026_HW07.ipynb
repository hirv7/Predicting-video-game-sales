**b. Take a look at the data dictionaries or metadata to understand the meaning of each variable.**

Make sure to differenciate between technical type and meaning (integers for categorical, strings for dates, etc).

Find out the allowed values or range for each variable, check how categories are coded, and missing values represented. Think of the role of the variable in the project (target, candidate predictor, identifier, date/time, or not relevant).

If you can, try to find where the documentation and the data disagree.

You can do this as a text paragraph, a list, or as a table suggested below.


```python
## TODO
rows = []

for col in data.columns:
    s = data[col]
    # values/range
    if pd.api.types.is_numeric_dtype(s):
        values = f"{s.min()} to {s.max()}"
    elif pd.api.types.is_datetime64_any_dtype(s):
        values = f"{s.min()} to {s.max()}"
    else:
        n_unique = s.nunique()
        if n_unique <= 10:
            values = ", ".join(map(str, s.dropna().unique()))
        else:
            values = f"{n_unique} unique values, e.g. " + ", ".join(map(str, s.dropna().unique()[:3]))

    rows.append({
        "variable": col,
        "type": str(s.dtype),
        "values/range": values,
        "missing coding": f"NaN ({s.isna().sum()} = {s.isna().mean():.1%})"        
    })

summary = pd.DataFrame(rows)
display(summary)
```

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## TODO
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
sns.scatterplot(data=data, x="critic_score", y="total_sales", alpha=0.3)
```

```python
top_consoles = data["console"].value_counts().head(15)

top_consoles.plot(kind="bar", figsize=(10, 5))
plt.title("Number of games for the top 15 consoles")
plt.xlabel("Console")
plt.ylabel("Number of games")
plt.show()
```

```python
data["release_date"] = pd.to_datetime(data["release_date"], errors="coerce")
data["year"] = data["release_date"].dt.year

sales_by_year = data.groupby("year")[["na_sales", "jp_sales", "pal_sales", "other_sales"]].sum()

sales_by_year.plot(kind="line", figsize=(10, 5))
plt.title("Total sales per year by region")
plt.xlabel("Release year")
plt.ylabel("Sales (millions of copies)")
plt.show()
```

```python
data["genre"].value_counts().sort_values().plot(kind="barh")
plt.title("Number of games per genre")
plt.xlabel("Number of games")
plt.ylabel("Genre")
plt.show()
```

**Answer:** 
- PC has multiple times more games than the next most popular console.
- Video game sales hit a peak around 2008-2010 and then started falling.
- The genre with the most games is "Misc".
